```julia
# Today's network: towns A, B, C (rows), facilities M, W, L (columns)
waste_today = [100, 80, 60]                 # [Mg/day]
collection_cost = [70 84 95;                # [$/Mg] town i to facility j
                   84 72 85;
                   98 96 65]
residual_cost = [65, 55]                    # [$/Mg] MRF residual to W, to L
ash_cost = 52.5                             # [$/Mg] WTE ash to L
recovery_fraction = 0.4                     # share of MRF waste recycled
ash_fraction = 0.25                         # share burned left as ash
capacity = [80, 70, 160]                    # [Mg/day] M, W, L
recycling_today = 0.1                       # share of the waste recycled

# The planning year
waste_planning = [135, 108, 81]             # [Mg/day]
recycling_planning = 0.25
emission_factor = [0, 0.2, 0.6]             # [t CO2e/Mg] at M, W, a landfill
emissions_limit = 96                        # [t CO2e/day]
# [truck-hours/Mg] from town i to M, W, L, N
truck_hours = [0.065 0.085 0.15 0.225;
               0.1 0.055 0.125 0.2;
               0.135 0.115 0.075 0.085]
fleet_hours = 24                            # [truck-hours/day]

# The candidate site N
distance_to_N = [40, 35, 12]                # [km] from towns A, B, C
mrf_to_N = 30                               # [km]
wte_to_N = 25                               # [km]
collection_rate = 2                         # [$/Mg-km]
trailer_rate = 0.5                          # [$/Mg-km]
operating_cost_N = 25                       # [$/Mg]

# Expansion options: MRF, WTE, landfill at N
fixed_cost = [1000, 1500, 2500]             # [$/day] if built
capacity_cost = [10, 60, 20]                # [$/day per Mg/day]
largest_size = [150, 80, 150]               # [Mg/day]
```

#### Problem 2.1 (5)

Implement today’s model in `JuMP` and solve it. If your model is right,

Provide the output flows as a table and check that the solution matches.

```julia
# Today's model, adapting lecture's waste_model
function waste_model(; landfill_capacity=capacity[3])
    cap = [capacity[1], capacity[2], landfill_capacity]
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    @variable(model, W[1:3, 1:3] >= 0)   # town i -> facility j [Mg/day]
    @variable(model, R_MW >= 0)          # MRF residual -> WTE
    @variable(model, R_ML >= 0)          # MRF residual -> LF
    @variable(model, R_WL >= 0)          # WTE ash -> LF
    @objective(model, Min,
        sum(collection_cost[i, j] * W[i, j] for i in 1:3, j in 1:3) +
        residual_cost[1] * R_MW + residual_cost[2] * R_ML + ash_cost * R_WL)
    @constraint(model, town[i in 1:3], sum(W[i, :]) == waste_today[i])
    @constraint(model, ash, R_WL == ash_fraction * (W[1, 2] + W[2, 2] + W[3, 2] + R_MW))
    @constraint(model, residual,
        R_MW + R_ML == (1 - recovery_fraction) * (W[1, 1] + W[2, 1] + W[3, 1]))
    @constraint(model, mrf, W[1, 1] + W[2, 1] + W[3, 1] <= cap[1])
    @constraint(model, wte, W[1, 2] + W[2, 2] + W[3, 2] + R_MW <= cap[2])
    @constraint(model, lf, W[1, 3] + W[2, 3] + W[3, 3] + R_ML + R_WL <= cap[3])
    @constraint(model, recycling,
        recovery_fraction * (W[1, 1] + W[2, 1] + W[3, 1]) >= recycling_today * sum(waste_today))
    return model
end

today = waste_model()
optimize!(today)
println("Cost: \$", round(objective_value(today), digits=2), "/day")

W = value.(today[:W])
flows = DataFrame(
    Flow = ["From Town A", "From Town B", "From Town C", "MRF residual", "WTE ash"],
    M = [W[1, 1], W[2, 1], W[3, 1], missing, missing],
    W = [W[1, 2], W[2, 2], W[3, 2], value(today[:R_MW]), missing],
    L = [W[1, 3], W[2, 3], W[3, 3], value(today[:R_ML]), value(today[:R_WL])],
)
```

#### Problem 2.2 (8)

Solve the new model, with the expansion options from Problem 1.2 and the
constraints from Problem 1.3. Report which options should be built and
their sizes, the daily cost, and the non-zero flows as a table. Compare
the emissions, the waste recycled and the truck-hours with their limits.
Which constraints are binding?

```julia
# cost of sending 1 Mg to N: hauling + N's operating cost
cost_to_N = collection_rate .* distance_to_N .+ operating_cost_N   # A, B, C
residual_to_N = trailer_rate * mrf_to_N + operating_cost_N
ash_to_N = trailer_rate * wte_to_N + operating_cost_N
cost = [collection_cost cost_to_N]          # columns M, W, L, N

# New model with the expansion options and the planning-year rules
function expansion_model(; with_emissions=true)
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    @variable(model, W[1:3, 1:4] >= 0)   # town i -> M, W, L, N [Mg/day]
    @variable(model, R_MW >= 0)          # MRF residual -> WTE
    @variable(model, R_ML >= 0)          # MRF residual -> LF
    @variable(model, R_MN >= 0)          # MRF residual -> N
    @variable(model, R_WL >= 0)          # WTE ash -> LF
    @variable(model, R_WN >= 0)          # WTE ash -> N
    @variable(model, X[1:3], Bin)        # expand MRF, expand WTE, build N
    @variable(model, K[1:3] >= 0)        # added at MRF, added at WTE, size of N [Mg/day]
    @objective(model, Min,
        sum(cost[i, j] * W[i, j] for i in 1:3, j in 1:4) +
        residual_cost[1] * R_MW + residual_cost[2] * R_ML + residual_to_N * R_MN +
        ash_cost * R_WL + ash_to_N * R_WN +
        sum(fixed_cost[k] * X[k] + capacity_cost[k] * K[k] for k in 1:3))
    @constraint(model, town[i in 1:3], sum(W[i, :]) == waste_planning[i])
    @constraint(model, ash, R_WL + R_WN == ash_fraction * (W[1, 2] + W[2, 2] + W[3, 2] + R_MW))
    @constraint(model, residual,
        R_MW + R_ML + R_MN == (1 - recovery_fraction) * (W[1, 1] + W[2, 1] + W[3, 1]))
    @constraint(model, mrf, W[1, 1] + W[2, 1] + W[3, 1] <= capacity[1] + K[1])
    @constraint(model, wte, W[1, 2] + W[2, 2] + W[3, 2] + R_MW <= capacity[2] + K[2])
    @constraint(model, lf, W[1, 3] + W[2, 3] + W[3, 3] + R_ML + R_WL <= capacity[3])
    @constraint(model, lf_N, W[1, 4] + W[2, 4] + W[3, 4] + R_MN + R_WN <= K[3])
    @constraint(model, max_size[k in 1:3], K[k] <= largest_size[k] * X[k])
    @constraint(model, recycling,
        recovery_fraction * (W[1, 1] + W[2, 1] + W[3, 1]) >= recycling_planning * sum(waste_planning))
    @constraint(model, hours, sum(truck_hours[i, j] * W[i, j] for i in 1:3, j in 1:4) <= fleet_hours)
    if with_emissions
        @constraint(model, emissions,
            emission_factor[2] * (W[1, 2] + W[2, 2] + W[3, 2] + R_MW) +
            emission_factor[3] * (sum(W[:, 3]) + sum(W[:, 4]) + R_ML + R_MN) <= emissions_limit)
    end
    return model
end

planned = expansion_model()
optimize!(planned)
println("Cost: \$", round(objective_value(planned), digits=2), "/day")

options = DataFrame(
    Option = ["Expand the MRF", "Expand the WTE", "Build a landfill at N"],
    Built = round.(Int, value.(planned[:X])) .== 1,
    Size = round.(value.(planned[:K]), digits=2),
)
display(options)

# non-zero flows
W = value.(planned[:W])
R = value.([planned[:R_MW], planned[:R_ML], planned[:R_MN], planned[:R_WL], planned[:R_WN]])
edges = [vec(["Town $t -> $f" for t in "ABC", f in ["M", "W", "L", "N"]]);
         ["MRF residual -> W", "MRF residual -> L", "MRF residual -> N", "WTE ash -> L", "WTE ash -> N"]]
flows = [vec(W); R]
keep = flows .> 1e-6
nonzero = DataFrame(Edge = edges[keep], Flow = round.(flows[keep], digits=2))
display(nonzero)

# compare with the limits
recycled = recovery_fraction * sum(W[:, 1])
em = emission_factor[2] * (sum(W[:, 2]) + value(planned[:R_MW])) + emission_factor[3] * (sum(W[:, 3:4]) + value(planned[:R_ML]) + value(planned[:R_MN]))
hrs = sum(truck_hours .* W)

limits = DataFrame(
    Rule = ["Recycling (Mg/day)", "Emissions (t CO2e/day)", "Truck-hours (h/day)"],
    Value = round.([recycled, em, hrs], digits=2),
    Limit = [recycling_planning * sum(waste_planning), emissions_limit, fleet_hours],
)
display(limits)
```
